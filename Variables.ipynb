The following code demonstrates the difference between standard division (`/`) and floor division (`//`):

*   **Standard Division (`/`)**: Returns a float, including any fractional part.
*   **Floor Division (`//`)**: Returns an integer if both operands are integers, or a float with `.0` if either operand is a float. It truncates the fractional part, effectively rounding down to the nearest whole number.

For example:

*   `10 / 3` results in `3.3333333333333335`.
*   `10 // 3` results in `3`.
*   `10 / 4` results in `2.5`.
*   `10 // 4` results in `2`.

```python
# Demonstrating the difference between division and floor division

num1 = 10
num2 = 3

# Standard division
result_division = num1 / num2
print(f"Standard Division ({num1} / {num2}): {result_division}")

# Floor division
result_floor_division = num1 // num2
print(f"Floor Division ({num1} // {num2}): {result_floor_division}")

num3 = 10
num4 = 4

# Standard division
result_division_2 = num3 / num4
print(f"Standard Division ({num3} / {num4}): {result_division_2}")

# Floor division
result_floor_division_2 = num3 // num4
print(f"Floor Division ({num3} // {num4}): {result_floor_division_2}")
```

```python
# Exercise 1: Temperature Converter
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Test the function
temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c}°C is equal to {temp_f}°F")
```

The `chr()` function takes an ASCII value as input and returns its corresponding character.

*   **Input**: An ASCII integer value.
*   **Output**: The character represented by that ASCII value.

```python
print(chr(65))
print(chr(65+32))
```

The `ord()` function is the inverse of `chr()`. It takes a single character as input and returns its corresponding ASCII integer value.

*   **Input**: A single character string.
*   **Output**: The ASCII integer value of the character.

```python
# Demonstrating the ord() function
print(ord('A'))
print(ord('a'))
print(ord('!'))

#Converting upppercase to lowecase uring ord and chr functions
print(chr(ord("A")+32))
print(chr(ord("B")+32))
print(chr(ord("a")-32))
print(chr(ord("b")-32))
```

The `id()` function in Python returns the identity of an object. This identity is an integer that is guaranteed to be unique and constant for that object during its lifetime. Think of it as the object's memory address.

*   **Input**: An object (variable, number, string, etc.).
*   **Output**: An integer representing the object's unique identity (memory address).

```python
a=10
id(a)
print(id(a))
```
